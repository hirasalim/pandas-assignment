```python
# deleteing row and column

import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Age": [20, 21, 19],
    "Marks": [80, 90, 75]
}

df = pd.DataFrame(data)
print(df)
# deleting column
print("===========================")
df = df.drop("Age", axis=1)
print(df)

#deleting row
print("===========================")

df=df.drop(1,axis=0)
print(df)
```

```python
# missong value
data = {
    "Name": ["Ali", "Aisha", "Bilal", "Sara", "Zain"],
    "Marks": [80, None, 75,None,50,]
}

df = pd.DataFrame(data)

print(df)
print("+=============================")
print(df.isnull())
print(df.isnull().sum())
# removing missing value
print(df.dropna())
print("=============================")
```

```python
# fill missing value
data = {
    "Name": ["Ali", "Aisha", "Bilal", "Sara", "Zain"],
    "Marks": [80, None, 75,None,50,]
}

df = pd.DataFrame(data)

print(df)
print("=============================")
df=df.fillna(1)
print(df)
# df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
```

```python
# duplicate values
data = {
    "Name": ["Ali", "Sara", "Ali"],
    "Marks": [80, 90, 80]
}

df = pd.DataFrame(data)
print(df)
print("=============================")
print(df.duplicated())
print("=============================")
print(df.duplicated().sum())
print("=============================")
print(df.drop_duplicates())
```

```python
# sorting value

data = {
    "Name": ["Zain", "Qadir", "Bilal", "Sara", "Aisha"],
    "Marks": [80, 56, 75,None,50,]
}

df = pd.DataFrame(data)

print(df)
print("=============================")

print(df.sort_values("Name", ascending=True))
```

```python
# Aggregation
data = {
    "Name": ["Zain", "Qadir", "Bilal", "Sara", "Aisha"],
    "marks": [80, 56, 75,None,50,]
}

df = pd.DataFrame(data)

print(df)
print("=============================")
print(df['marks'].sum())
print(df['marks'].mean())
print(df['marks'].max())
print(df['marks'].min())
print(df['marks'].count())
```

```python
#groupby
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Course": ["AI", "Web", "AI", "Web"],
    "Marks": [80, 90, 70, 85]
}

df = pd.DataFrame(data)
result = df.groupby("Course")["Marks"].mean()

print(result)
```

```python
# combine data
df1 = pd.DataFrame({
    "Name": ["Ali", "Sara"],
    "Marks": [80, 90]
})

df2 = pd.DataFrame({
    "Name": ["Ahmed", "Ayesha"],
    "Marks": [75, 85]
})
final_df = pd.concat([df1, df2], ignore_index=True)

print(final_df)
```

```python
# merg data
students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ali", "Sara", "Ahmed"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [80, 90, 75]
})
final_df = pd.merge(students, marks, on="ID")

print(final_df)
```
